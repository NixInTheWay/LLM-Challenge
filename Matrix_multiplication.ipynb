```python
def create_matrix(Row,Column):
  matrix=[]
  for i in range(Row):
    matrix.append([])
    for j in range(Column):
      num = int(input("enter a number"))
      matrix[i].append(num)
  print("Matrix created")
  return matrix
```

```python
def multyply_matrix(A,B):
  A_rows = len(A)
  B_rows = len(B)
  A_columns = len(A[0])
  B_columns = len(B[0])
  if A_columns != B_rows:
    raise ValueError("Can't multiply dissimilair matrices")
    return
  Row = A_rows
  Column = B_columns
  matrix=[]
  for i in range(Row):
    matrix.append([])
    for j in range(Column):
      matrix[i].append(0)
  for i in range(Row):
    for j in range(Column):
      total_sum = 0
      for element in range(A_columns):
        prod = A[i][element]*B[element][j]
        total_sum += prod
      matrix[i][j] = total_sum
  return matrix
```

```python
A = create_matrix(2,2)
B = create_matrix(2,2)
result = multyply_matrix(A,B)
for i in result:
  print(i)
```
